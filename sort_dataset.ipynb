```python
import pandas as pd

# Employee Dataset (sample)
df = pd.DataFrame({
    "Name": ["Arun", "Divya", "Karthik", "Meena", "Ravi", "Sneha"],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "IT"],
    "Age": [28, 34, 25, 41, 30, 28],
    "Salary": [52000, 48000, 61000, 75000, 45000, 58000]
})

# 1. Ascending sort (numerical column)
asc = df.sort_values(by="Salary")
print("Salary - Ascending:\n", asc, "\n")

# 2. Descending sort (numerical column)
desc = df.sort_values(by="Salary", ascending=False)
print("Salary - Descending:\n", desc, "\n")

# 3. Categorical column sort (A to Z, then Z to A)
print(df.sort_values(by="Department"), "\n")
print(df.sort_values(by="Department", ascending=False), "\n")

# 4. Multiple columns: Department A-Z, then Salary high to low
multi = df.sort_values(by=["Department", "Salary"], ascending=[True, False])
print("Multi-column sort:\n", multi, "\n")

# 5. Reset index for a clean output
print(multi.reset_index(drop=True))
```
